todo-list-priority-manager

```python
from datetime import datetime, timedelta
import random

class Task:
    def __init__(self, title, priority='medium', category='general', due_days=7):
        self.id = random.randint(1000, 9999)
        self.title = title
        self.priority = priority
        self.category = category
        self.due = (datetime.now() + timedelta(days=due_days)).strftime('%Y-%m-%d')
        self.done = False
        self.created = datetime.now().strftime('%Y-%m-%d %H:%M')

    def complete(self):
        self.done = True

class TodoManager:
    priority_order = {'high': 0, 'medium': 1, 'low': 2}

    def __init__(self):
        self.tasks = []

    def add(self, title, priority='medium', category='general', due_days=7):
        t = Task(title, priority, category, due_days)
        self.tasks.append(t)
        print(f"✅ Added: [{t.id}] {t.title}")
        return t

    def complete(self, task_id):
        for t in self.tasks:
            if t.id == task_id:
                t.complete()
                print(f"🎉 Completed: {t.title}")

    def show(self, filter_done=False):
        tasks = [t for t in self.tasks if t.done == filter_done]
        tasks.sort(key=lambda x: self.priority_order.get(x.priority, 1))
        icons = {'high':'🔴','medium':'🟡','low':'🟢'}
        status = "✅ Done" if filter_done else "⏳ Pending"
        print(f"\n===== {status} Tasks ({len(tasks)}) =====")
        for t in tasks:
            icon = icons.get(t.priority, '⚪')
            print(f"{icon} [{t.id}] {t.title:<30} | {t.category:<12} | Due: {t.due}")

tm = TodoManager()
t1 = tm.add("Submit project report", "high", "work", 2)
t2 = tm.add("Buy groceries", "medium", "personal", 1)
t3 = tm.add("Read Python book", "low", "learning", 14)
t4 = tm.add("Fix production bug", "high", "work", 0)
t5 = tm.add("Call dentist", "medium", "health", 3)
tm.complete(t2.id)
tm.show(filter_done=False)
tm.show(filter_done=True)
```
